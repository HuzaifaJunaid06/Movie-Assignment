### **Question 34**

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('cleaned_movies.csv')
df['Gross'] = pd.to_numeric(df['Gross'], errors='coerce')
df['Genre_Group'] = df['Genre'].apply(lambda x: 'Drama' if 'Drama' in str(x) else ('Action' if 'Action' in str(x) else 'Other'))
filtered_df = df[df['Genre_Group'].isin(['Drama', 'Action'])]
genre_gross = filtered_df.groupby('Genre_Group')['Gross'].sum()
genre_gross.plot(kind='bar', color=['dodgerblue', 'orange'], width=0.5)
plt.title('Drama vs Action Total Gross Revenue')
plt.xlabel('Genre Group')
plt.ylabel('Total Gross')
plt.xticks(rotation=0)
plt.show()
```

### **Question 38**
Un movies ke naam aur ratings batayein jinka 'Released_Year' 2000 se pehle ka hai AUR unka 'Gross' unke overall median gross se zyada hai.

```python
import pandas as pd
df = pd.read_csv('cleaned_movies.csv')
df['Released_Year'] = pd.to_numeric(df['Released_Year'], errors='coerce')
df['Gross'] = pd.to_numeric(df['Gross'], errors='coerce')
df['IMDB_Rating'] = pd.to_numeric(df['IMDB_Rating'], errors='coerce')
median_gross = df['Gross'].median()
condition = (df['Released_Year'] < 2000) & (df['Gross'] > median_gross)
result = df[condition][['Series_Title', 'IMDB_Rating', 'Released_Year', 'Gross']]
print("Movies before 2000 with Gross > Median Gross:")
display(result)
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('cleaned_movies.csv')
df['IMDB_Rating'] = pd.to_numeric(df['IMDB_Rating'], errors='coerce')
df['Meta_score'] = pd.to_numeric(df['Meta_score'], errors='coerce')
top_5 = df.sort_values(by='IMDB_Rating', ascending=False).head(5)
top_5.set_index('Series_Title')[['IMDB_Rating', 'Meta_score']].plot(kind='bar', figsize=(10, 6), width=0.8)
plt.title('Top 5 Movies: IMDB Rating (Scaled to 100) vs Meta_score')
plt.xlabel('Movie Title')
plt.ylabel('Score')
plt.show()
```
